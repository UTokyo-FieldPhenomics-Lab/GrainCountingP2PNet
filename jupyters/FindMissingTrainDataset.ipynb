```python
import os
from glob import glob
from pathlib import Path

import pandas as pd
```

```python
old_dataset = "../data/demo_dataset.old"
new_dataset = "../data/demo_dataset"
```

```python
# Get all files in the directory
old_files = glob(f'{old_dataset}/images/train/*')

print(len(old_files))

# Extract unique prefixes
old_prefixes = list(Path(f).stem.split('_v')[0] for f in old_files)

old_unique_prefixes = {k: old_prefixes.count(k) for k in set(old_prefixes)}

print(old_unique_prefixes)
len(old_unique_prefixes.keys())
```

```python
Path(old_files[0]).stem
```

```python
Path(old_files[0]).stem.split('_v')[0]
```

```python
# Get all files in the directory
new_files = glob(f'{new_dataset}/images/train/*')

print(len(new_files))

# Extract unique prefixes
new_prefixes = list(Path(f).stem.split('_x')[0] for f in new_files)

new_unique_prefixes = {k: new_prefixes.count(k) for k in set(new_prefixes)}

print(new_unique_prefixes)
len(set(new_prefixes))
```

```python
compare = pd.DataFrame(columns=['old', 'new'])

for key in old_unique_prefixes.keys():
    compare.loc[key, 'old'] = old_unique_prefixes[key]
    if key in new_unique_prefixes.keys():
        compare.loc[key, 'new'] = new_unique_prefixes[key]
    else:
        compare.loc[key, 'new'] = -1

compare
```

```python
compare.to_csv('outputs/dataset_compare.csv')
```
